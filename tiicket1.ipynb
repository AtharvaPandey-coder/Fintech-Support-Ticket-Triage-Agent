```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_selection import chi2
import pandas as pd
```

```python
df['clean_text']=df['Consumer complaint narrative'].astype(str).str.lower().str.replace(r'[^a-z0-9\s]','',regex=True).str.strip()
```

```python
cv=CountVectorizer()
X_counts = cv.fit_transform(df['clean_text'])
y_labels = df['Product']
```

```python
chi_scores,p_value=chi2(X_counts,y_labels)
```

```python
results=pd.DataFrame({
    'word':cv.get_feature_names_out(),
    'chi2_scores':chi_scores,
    'p_value':p_value
}).sort_values('chi2_scores',ascending=False)

print(results.head(20))
```

```python
significant=(results['p_value'] < 0.05).sum()
total=len(results)
print(f"Significant words :{significant}/{total} ({significant/total*100:.1f}%)")
```

```python
X=df['clean_text']
y=df['Product_final']

X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=42,test_size=0.2,stratify=y)
```

```python
def keyword_baseline(text):
    text=text.lower()
    if any(w in text for w in ['mortage','escrow','forclose']):
        return 'Mortgage'
    elif any(w in text for w in ['credit card','card']):
        return 'Credit card'
    elif any(w in text for w in ['debt collector','debt','owed']):
        return 'Debt collection'
    elif any(w in text for w in ['credit report','equifax','experian','transunion']):
        return 'Credit reporting'
    elif any(w in text for w in ['bank report','checking','atm','debit']):
        return 'Bank account or service'
    elif any(w in text for w in ['student loan','navient','sallie mae']):
        return 'Student loan'
    elif any(w in text for w in ['payday']):
        return 'Payday loan'
    elif any (w in text for w in ['prepaid']):
        return 'Prepaid card'
    elif any(w in text for w in ['wire transfer','money transfer','remittance']):
        return 'Money transfers'
    elif any(w in text for w in ['auto loan','car loan','vehicle loan']):
        return 'Consumer Loan'
    else:
        return 'Other'
```

```python
baseline_preds=X_test.apply(keyword_baseline)
print(classification_report(y_test,baseline_preds))
```

```python
#3 Now we will use our method b which is the trained model
tfidf=TfidfVectorizer(stop_words='english',ngram_range=(1,2),max_features=5000)
tfidf_train=tfidf.fit_transform(X_train)
tfidf_test=tfidf.transform(X_test)
```

```python
lr=LogisticRegression()
lr.fit(tfidf_train,y_train)
lr_preds=lr.predict(tfidf_test)

print(classification_report(y_test,lr_preds))
```

```python
from sklearn.svm import LinearSVC
from sklearn.ensemble import RandomForestClassifier
svc=LinearSVC(class_weight='balanced',max_iter=2000)
svc.fit(tfidf_train,y_train)
svc_preds=svc.predict(tfidf_test)
print(classification_report(y_test,svc_preds))
```

```python
import os 
from groq import Groq
import pandas as pd
import time
from sklearn.metrics import classification_report
client=Groq(api_key=os.environ.get('GROQ_API_KEY'))

#3 small sample because llm calls take time/tokens
sample_df=pd.DataFrame({'text':X_test,'true_labels':y_test}).sample(100,random_state=42)

categories=df['Product_final'].unique().tolist()

def llm_classifier(text):
    prompt=f"""Classifiy the consumer complaint into exactly one of these categories {categories}
    Complaint:"{text[:500]}"
    Reply With only the catgeory name,nothing else"""
    response=client.chat.completions.create(
        model='openai/gpt-oss-20b',
        messages=[{'role':'user','content':prompt}],
        temperature=0      
    )
    time.sleep(2)
    return response.choices[0].message.content.strip()
```

```python
sample_df['llm_pred']=sample_df['text'].apply(llm_classifier)
print(classification_report(sample_df['true_labels'],sample_df['llm_pred'],zero_division=0))
```

```python
def sentiment_label(score):
    if score <= -0.5:
        return 'Angry'
    elif score <= 0:
        return 'Frustrated'
    else:
        return 'Calm/Neutral'

    
```
